```python
import numpy as np
np.random.seed(42)
print('Library imported successfully!')
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1],
])
y = np.array([0, 1, 1, 1])
```

```python
epochs = 10
```

```python
def step_activation(x, threshold=0.5):
    return 1 if x >= threshold else 0
```

```python
w = np.array([0.5, 2.0])
```

```python
error = 0
```

```python
def update(i, error, learning_rate=0.9):
    w[1] = w[1] + learning_rate * error * X[i][1]
    w[0] = w[0] + learning_rate * error * X[i][0]
```

```python
for epoch in range(epochs):
    errors = 0

    for i in range(len(X)):
        z = np.dot(X[i], w)
        pred = step_activation(z)

        error = y[i] - pred

        if error != 0:
            update(i, error)
            errors += 1

    print(f"Epoch {epoch + 1}: Errors = {errors}")

    if errors == 0:
        print("Training converged!")
        break

print("\nFinal weights:", w)
```
